```python
import numpy as np
import scipy as sp

from scipy.integrate import dblquad
import math

def f(y, x):
    return 1.0 / (np.sqrt(x + y) * (1.0 + x + y)**2)

val, err = dblquad(
    f,
    0.0, 1.0,
    # x: 0 -> 1
    lambda x: 0.0,
    # y 下限
    lambda x: 1.0 - x
    # y 上限
)
print("積分結果 =", val)
print("誤差 =", err)
```
