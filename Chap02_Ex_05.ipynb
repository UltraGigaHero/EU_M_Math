```python
import numpy as np
import scipy.linalg as linalg

matrix = np.array([[1,2,3], [1,3,2], [3,1,2]])
eig_value, eig_vector = linalg.eig(matrix)


print('固有値')
print(eig_value)
print('固有ベクトル')
print(eig_vector)
```
