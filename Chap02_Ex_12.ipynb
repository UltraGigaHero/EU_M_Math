```python
import numpy as np
import matplotlib.pyplot as plt

n = 10000

x = np.random.uniform(0.0,1.0,n)
y = np.random.uniform(0.0,1.0,n)

plt.figure(figsize = (20,6))

plt.subplot(1, 2, 1)
plt.hist(x, bins=60, range=(0, 1))
plt.title("x")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.hist(y, bins=60, range=(0, 1))
plt.title("y")
plt.grid(True)
```
