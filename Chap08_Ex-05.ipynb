```python
from sklearn.tree import DecisionTreeClassifier

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

datab = load_breast_cancer()

X = datab['data']
y = datab['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=0)

model = DecisionTreeClassifier(criterion = 'entropy', max_depth = 5, random_state = 0)
#木の深きは５です
model.fit(X_train, y_train)

print('スコア(train):{:.3f}'.format(model.score(X_train, y_train)))
print('スコア(test):{:.3f}'.format(model.score(X_test, y_test)))
```

```python
from sklearn.tree import DecisionTreeClassifier

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

datab = load_breast_cancer()

X = datab['data']
y = datab['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=0)

model = DecisionTreeClassifier(criterion = 'entropy', max_depth = 50, random_state = 0)
#木の深きは50です
model.fit(X_train, y_train)

print('スコア(train):{:.3f}'.format(model.score(X_train, y_train)))
print('スコア(test):{:.3f}'.format(model.score(X_test, y_test)))
```
