```python
from scipy.optimize import fsolve
import numpy as np

import matplotlib.pyplot as plt

def f(x):
   y = x**3 - 2 * x**2 - 10 * x + 12
   return y

x = np.linspace(-10,10)
plt.plot(x,f(x))
plt.plot(x,np.zeros(len(x)))
plt.grid(True)
```

```python
import matplotlib.pyplot as plt

def f(x):
   y = x**3 - 2 * x**2 - 10 * x + 12
   return y

x1 = fsolve(f,-2.5)
x2 = fsolve(f,1.25)
x3 = fsolve(f,3.75)
print('解：',x1,x2,x3)
```
