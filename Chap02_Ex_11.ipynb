```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-10, 10, 0.1)

y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure(figsize=(20, 6))
plt.plot(x, y_sin, label="y = sin(x)")
plt.plot(x, y_cos, label="y = cos(x)")
plt.grid(True)
```
