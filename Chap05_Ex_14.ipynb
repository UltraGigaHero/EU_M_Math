```python
import numpy as np
import scipy as sp

A = np.identity(3)
print(A)
A[0,:] = 1
A[:,0] = 1
A[0,0] = 3
b = np.ones(3)

(LU,piv) = sp.linalg.lu_factor(A)
L = np.identity(3) + np.tril(LU,-1)
U = np.triu(LU)
P =np.identity(3)[piv]

x = sp.linalg.lu_solve((LU,piv),b)

print('LU分解の結果:',x)
```
