```python
import numpy as np
import math

n = 10000

x = np.random.uniform(-1,1,n)
y = np.random.uniform(-1,1,n)

incircle = np.hypot(x,y) <= 1
#以上はChap02_CM_02のコードです

pi = 4 * incircle.sum() / n

print(pi)
```
