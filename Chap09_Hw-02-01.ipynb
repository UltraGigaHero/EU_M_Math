# k-means、k-means++、k-medoidsについて

## 1. k-means

k-meansは、データを指定した数のクラスタに分ける代表的なクラスタリング手法である。  
各クラスタの中心（重心）と各データ点との距離が小さくなるように、クラスタを繰り返し更新する。

### 実行方法

1. クラスタ数 `k` を決める
2. 初期の重心をランダムに選ぶ
3. 各データ点を最も近い重心のクラスタに割り当てる
4. 各クラスタの重心を再計算する
5. 割り当てが変わらなくなるまで 3 と 4 を繰り返す

### メリット

- アルゴリズムが単純で理解しやすい
- 計算速度が速い
- 大規模データにも適用しやすい
- scikit-learnなどで簡単に実装できる

### デメリット

- クラスタ数 `k` を事前に決める必要がある
- 初期重心の選び方によって結果が変わる
- 外れ値の影響を受けやすい
- 球状でないクラスタには弱い
- カテゴリデータにはそのまま使いにくい

## 2. k-means++

### 実行方法

1つ目の重心をランダムに選ぶ

1.   1つ目の重心をランダムに選ぶ
2.   既に選ばれた重心から遠いデータ点ほど、次の重心として選ばれやすくする
3.   k 個の重心が選ばれるまで繰り返す
4.   その後は通常のk-meansと同じ手順でクラスタリングする

### メリット

- 通常のk-meansより結果が安定しやすい
- 初期値の悪さによる失敗を減らせる
- 収束が速くなることがある
- scikit-learnのKMeansではデフォルトで利用される

### デメリット

- 通常のk-meansより初期化に少し時間がかかる
- 外れ値の影響は依然として受ける
- クラスタ数 k は事前に決める必要がある
- 球状でないクラスタには弱い

## 3. k-medoids

### 実行方法

1. クラスタ数 k を決める
2. 初期のmedoidを選ぶ
3. 各データ点を最も近いmedoidのクラスタに割り当てる
4. 各クラスタ内で、他の点との距離の合計が最も小さい点を新しいmedoidにする
5. medoidが変わらなくなるまで繰り返す

### メリット

- 外れ値の影響を受けにくい
- クラスタ中心が実際のデータ点なので解釈しやすい
- 任意の距離尺度を使いやすい
- k-meansより頑健なクラスタリングができる

### デメリット

- k-meansより計算コストが高い
- 大規模データでは時間がかかりやすい
- クラスタ数 k を事前に決める必要がある
- scikit-learn本体には標準実装がないため、追加ライブラリが必要になることがある
